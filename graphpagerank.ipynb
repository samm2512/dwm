```python
import numpy as np
import networkx as nx
```

```python
# Create Graph
G = nx.DiGraph([
    (1,2), (1,3), (1,4),
    (2,3), (2,4),
    (3,1),
    (4,1), (4,3)
])
```

```python
# Create Adjacency Matrix
A = np.array([
    [0,   0,   1,   0.5],
    [1/3, 0,   0,   0  ],
    [1/3, 0.5, 0,   0.5],
    [1/3, 0.5, 0,   0  ]
])

n = A.shape[0]

print("Matrix A:\n", A)
print("Column sums:", A.sum(axis=0))
```

```python

# PageRank Without Damping Factor
# Formula: Vk = A * Vk-1

def pagerank(A, max_iter=100, tol=1e-8):
    n = A.shape[0]
    v = np.ones(n) / n

    print("V0 =", v)

    for i in range(1, max_iter + 1):
        v_new = A @ v

        if np.allclose(v_new, v, atol=tol):
            return v_new, i

        v = v_new

    return v, max_iter

v, k = pagerank(A)

print("PageRank vector:", v)
print("Iterations:", k)
```
