# Experiment 9: Diabetes Classification Using Naive Bayes
### Gaussian Naive Bayes — Without Using Any Libraries

**Dataset:** Pima Indians Diabetes sample dataset

**Features:** Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin, BMI, DiabetesPedigreeFunction, Age  
**Target:** Outcome (`0` = No Diabetes, `1` = Diabetes)

> This notebook uses pure Python only. No pandas, NumPy, or scikit-learn is used.

## 1. Create Dataset

```python
data = [
    [1, 85, 66, 29, 0, 26.6, 0.351, 31, 0],
    [8, 183, 64, 0, 0, 23.3, 0.672, 32, 1],
    [1, 89, 66, 23, 94, 28.1, 0.167, 21, 0],
    [0, 137, 40, 35, 168, 43.1, 2.288, 33, 1],
    [5, 116, 74, 0, 0, 25.6, 0.201, 30, 0],
    [3, 78, 50, 32, 88, 31.0, 0.248, 26, 1],
    [10, 115, 0, 0, 0, 35.3, 0.134, 29, 0],
    [2, 197, 70, 45, 543, 30.5, 0.158, 53, 1],
    [4, 110, 92, 0, 0, 37.6, 0.191, 30, 0],
    [7, 196, 90, 0, 0, 39.8, 0.451, 41, 1]
]

# Column names for understanding the dataset
columns = [
    "Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
    "Insulin", "BMI", "DiabetesPedigreeFunction", "Age", "Outcome"
]

print("PIMA INDIANS DIABETES DATASET")
print("-" * 130)
print("Row".ljust(6), end="")
for col in columns:
    print(col.ljust(25), end="")
print()
print("-" * 130)

for i, row in enumerate(data, start=1):
    print(str(i).ljust(6), end="")
    for value in row:
        print(str(value).ljust(25), end="")
    print()
print("-" * 130)
```

## 2. Split Data into Training and Testing

```python
train_data = data[:7]
test_data = data[7:]

print("Training records:", len(train_data))
print("Testing records:", len(test_data))
```

## 3. Separate Data Based on Class

```python
def separate_by_class(data):
    separated = {}

    for row in data:
        label = row[-1]

        if label not in separated:
            separated[label] = []

        separated[label].append(row[:-1])

    return separated
```

## 4. Calculate Mean

```python
def mean(values):
    return sum(values) / len(values)
```

## 5. Calculate Variance

```python
def variance(values):
    m = mean(values)
    return sum((x - m) ** 2 for x in values) / len(values)
```

## 6. Gaussian Probability

```python
def gaussian_probability(x, m, v):
    if v == 0:
        return 1 if x == m else 1e-10

    exponent = -((x - m) ** 2) / (2 * v)
    return (1 / ((2 * 3.14159 * v) ** 0.5)) * (2.71828 ** exponent)
```

## 7. Train the Model

```python
def train(data):
    separated = separate_by_class(data)
    model = {}

    for label, rows in separated.items():
        model[label] = []

        for column in zip(*rows):
            m = mean(column)
            v = variance(column)
            model[label].append((m, v))

    return model
```

## 8. Predict Class

```python
def predict(model, input_data, total):
    probabilities = {}

    for label, features in model.items():
        count = len(model_data[label])
        probability = count / total

        for i in range(len(features)):
            m, v = features[i]
            p = gaussian_probability(input_data[i], m, v)
            probability *= p

        probabilities[label] = probability

    return max(probabilities, key=probabilities.get)
```

## 9. Train Model

```python
model_data = separate_by_class(train_data)
model = train(train_data)

print("MODEL TRAINED SUCCESSFULLY!")
```

## 10. Display Mean and Variance

```python
print("MEAN AND VARIANCE")

for label, features in model.items():
    print("\nClass:", label)

    for i, (m, v) in enumerate(features):
        print("Feature", i + 1,
              "Mean:", round(m, 2),
              "Variance:", round(v, 2))
```

## 11. Test the Model

```python
y_actual = []
y_predicted = []

for row in test_data:
    actual = row[-1]
    prediction = predict(model, row[:-1], len(train_data))

    y_actual.append(actual)
    y_predicted.append(prediction)

    print("Actual:", actual, "Predicted:", prediction)
```

## 12. Calculate Accuracy

```python
correct = 0

for i in range(len(y_actual)):
    if y_actual[i] == y_predicted[i]:
        correct += 1

accuracy = correct / len(y_actual)
print("ACCURACY:", round(accuracy * 100, 2), "%")
```

## 13. Confusion Matrix

```python
TP = 0
TN = 0
FP = 0
FN = 0

for i in range(len(y_actual)):
    if y_actual[i] == 1 and y_predicted[i] == 1:
        TP += 1
    elif y_actual[i] == 0 and y_predicted[i] == 0:
        TN += 1
    elif y_actual[i] == 0 and y_predicted[i] == 1:
        FP += 1
    else:
        FN += 1

print("CONFUSION MATRIX")
print("TP:", TP)
print("TN:", TN)
print("FP:", FP)
print("FN:", FN)
```

## 14. Predict New Patient

```python
new_patient = [2, 120, 70, 25, 80, 28.5, 0.5, 30]

result = predict(model, new_patient, len(train_data))

print("NEW PATIENT:", new_patient)

if result == 1:
    print("PREDICTION: DIABETES")
else:
    print("PREDICTION: NO DIABETES")
```
