```python
import pandas as pd

df = pd.read_csv('/content/GroceryStoreDataSet.csv')

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())

print("\nColumn Names:")
print(df.columns)
```

```python
transactions = []

for row in df.iloc[:, 0]:
    items = [item.strip() for item in row.split(',')]
    transactions.append(items)

print("Number of Transactions:", len(transactions))

print("\nTransactions:")
for i, transaction in enumerate(transactions, 1):
    print(i, transaction)
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

basket = pd.DataFrame(
    te_array,
    columns=te.columns_
)

print("One-Hot Encoded Dataset:")
print(basket)
```

```python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(
    basket,
    min_support=0.20,
    use_colnames=True
)

print("Frequent Itemsets:")
print(frequent_itemsets)
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.80
)

print("Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
print("Strong Association Rules:")
print("-" * 60)

for _, row in rules.iterrows():
    print(
        set(row['antecedents']),
        "->",
        set(row['consequents']),
        "| Support =", round(row['support'], 4),
        "| Confidence =", round(row['confidence'] * 100, 2), "%",
        "| Lift =", round(row['lift'], 4)
    )
```

```python
import numpy as np
from scipy.stats import chi2_contingency

antecedent = 'MAGGI'
consequent = 'TEA'

a = b = c = d = 0

for transaction in transactions:
    has_maggi = antecedent in transaction
    has_tea = consequent in transaction

    if has_maggi and has_tea:
        a += 1
    elif has_maggi and not has_tea:
        b += 1
    elif not has_maggi and has_tea:
        c += 1
    else:
        d += 1

contingency_table = np.array([
    [a, b],
    [c, d]
])

chi_square, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Rule:", antecedent, "->", consequent)
print("\nContingency Table:")
print(contingency_table)

print("\nChi-Square =", round(chi_square, 4))
```

```python
confidence_maggi_tea = 4 / 5
confidence_tea_maggi = 4 / 7

max_confidence = max(
    confidence_maggi_tea,
    confidence_tea_maggi
)

print("Rule: MAGGI -> TEA")
print("Max Confidence =", round(max_confidence, 4))
```

```python
import math

# Actual values from the dataset
support_maggi = 5 / len(transactions)
support_tea = 7 / len(transactions)
support_maggi_tea = 4 / len(transactions)

# Confidence
confidence_maggi_tea = support_maggi_tea / support_maggi
confidence_tea_maggi = support_maggi_tea / support_tea

# All Confidence
all_confidence = support_maggi_tea / max(
    support_maggi,
    support_tea
)

# Max Confidence
max_confidence = max(
    confidence_maggi_tea,
    confidence_tea_maggi
)

# Kulc
kulc = (
    confidence_maggi_tea +
    confidence_tea_maggi
) / 2

# Cosine
cosine = support_maggi_tea / math.sqrt(
    support_maggi * support_tea
)

print("Pattern Evaluation Measures")
print("-" * 50)

print("Rule              : MAGGI -> TEA")
print("Support           :", round(support_maggi_tea, 4))
print("Confidence        :", round(confidence_maggi_tea, 4))
print("All Confidence    :", round(all_confidence, 4))
print("Max Confidence    :", round(max_confidence, 4))
print("Kulc              :", round(kulc, 4))
print("Cosine            :", round(cosine, 4))
```
