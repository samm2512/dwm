```python
import numpy as np

transactions = np.array([
    ['A', 'C', 'D', '', ''],
    ['B', 'C', 'E', '', ''],
    ['A', 'B', 'C', 'E', ''],
    ['B', 'E', '', '', '']
])

min_support = 0.50
min_confidence = 0.80

print("Transaction Database:")
for i, transaction in enumerate(transactions, 100):
    print(i, [str(x) for x in transaction if x != ''])

print("\nMinimum Support:", min_support)
print("Minimum Confidence:", min_confidence)
```

```python
items = np.unique(transactions[transactions != ''])

print("Unique Items:")

for item in items:
    print(str(item))
```

```python

support_count = {}
support = {}

print("C1 Support:")

for item in items:
    count = 0

    for transaction in transactions:
        if np.any(transaction == item):
            count += 1

    support_value = count / len(transactions)

    support_count[str(item)] = count
    support[str(item)] = support_value

    print(
        "{" + str(item) + "}",
        "Support Count =", count,
        "Support =", round(support_value, 2)
    )

L1 = []

for item in items:
    if support[str(item)] >= min_support:
        L1.append(str(item))

print("\nFrequent 1-Itemsets (L1):")

for item in L1:
    print("{" + item + "}")
```

```python
C2 = np.array([
    ['A', 'B'],
    ['A', 'C'],
    ['A', 'E'],
    ['B', 'C'],
    ['B', 'E'],
    ['C', 'E']
])

print("Candidate 2-Itemsets (C2):")

for itemset in C2:
    a = str(itemset[0])
    b = str(itemset[1])
    print("{" + a + ", " + b + "}")
```

```python
L2 = []

print("C2 Support:")

for itemset in C2:

    count = 0

    for transaction in transactions:
        if np.all(np.isin(itemset, transaction)):
            count += 1

    support_value = count / len(transactions)

    a = str(itemset[0])
    b = str(itemset[1])

    print(
        "{" + a + ", " + b + "}",
        "Support Count =", count,
        "Support =", round(support_value, 2)
    )

    if support_value >= min_support:
        L2.append([a, b])

print("\nFrequent 2-Itemsets (L2):")

for itemset in L2:
    print("{" + ", ".join(itemset) + "}")
```

```python
C3 = np.array([
    ['A', 'B', 'C'],
    ['A', 'C', 'E'],
    ['B', 'C', 'E']
])

print("Candidate 3-Itemsets (C3):")

for itemset in C3:
    print("{" + ", ".join(str(x) for x in itemset) + "}")
```

```python
L3 = []

print("\nC3 Support:")

for itemset in C3:

    count = 0

    for transaction in transactions:
        if np.all(np.isin(itemset, transaction)):
            count += 1

    support_value = count / len(transactions)

    print(
        "{" + ", ".join(str(x) for x in itemset) + "}",
        "Support Count =", count,
        "Support =", round(support_value, 2)
    )

    if support_value >= min_support:
        L3.append([str(x) for x in itemset])

print("\nFrequent 3-Itemsets (L3):")

for itemset in L3:
    print("{" + ", ".join(itemset) + "}")
```

```python
from itertools import combinations

print("Strong Association Rules:")
print("-" * 50)

all_frequent = L2 + L3

for itemset in all_frequent:

    itemset = [str(x) for x in itemset]

    if len(itemset) < 2:
        continue

    for r in range(1, len(itemset)):

        for antecedent_tuple in combinations(itemset, r):

            antecedent = list(antecedent_tuple)
            consequent = [
                item for item in itemset
                if item not in antecedent
            ]

            antecedent_count = 0
            itemset_count = 0

            for transaction in transactions:

                if np.all(np.isin(antecedent, transaction)):
                    antecedent_count += 1

                if np.all(np.isin(itemset, transaction)):
                    itemset_count += 1

            confidence = itemset_count / antecedent_count

            support_value = itemset_count / len(transactions)

            if confidence >= min_confidence:

                left = "{" + ", ".join(antecedent) + "}"
                right = "{" + ", ".join(consequent) + "}"

                print(
                    left, "->", right,
                    "Support =", round(support_value, 2),
                    "Confidence =", round(confidence * 100, 2), "%"
                )
```
